The following example shows how to generate training data. We want to train the drone flying in the y-z plane at different wall distances.

The tuning variables / settings live in `config.yaml` next to this notebook. This notebook is only a launcher; running the integrated project from the command line is more stable than running it inside the notebook.

Notes:
- Turn off disturbance observer is not a requirement. Any given controller output can correspond to a disturbance force.
- Run the scenario in simulation_user_guide.ipynb first to make sure the drone does not hit the wall. Adjust wall distance from performance.
- Use the same trajectory in every condition as the same setting in Neural Fly paper.
- Use RandomWaypointsInConstrainedSpace to fly in a constrained 3D space in front of a wall.
- When training disturbance model with inflow model in controller, enable inflow model option in controller.
- Use baseline disturbance observer to generate training data.
- Ensure the disturbance model is set to the correct condition, i.e. setting of sinusoidal wind.
- Restart the notebook before running.
