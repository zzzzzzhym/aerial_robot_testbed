```python
import vtk_reader
from pathlib import Path

import vtk_plotter as plotter

parameterized_folder = Path(r"tbd")

# base_path = Path(r"C:\Users\jiexu\Downloads\Aerocae Robotics - Geng 2025\export") / parameterized_folder
base_path = Path(r"C:\Users\jiexu\Downloads\Aerocae Robotics - Geng 2025\export")

reader = vtk_reader.VtkReader(base_path)
reader.load_mesh_by_wind_velocity([-5, 0, 0])

# 2. Create plotters
plotter_slices = plotter.plot_slices(
    reader.mesh,
    x_stride=10,
    y_stride=10,
    z_stride=10,
    arrow_factor=0.05,
)

plotter_3d_v = plotter.plot_velocity_arrows(
    reader.mesh,
    factor=0.05,
    x_stride=5,
    y_stride=5,
    z_stride=50,
)

# 3. Show both windows
plotter_slices.show()                # blocking — keeps the app alive
plotter_3d_v.show()

plotter.plot_velocity_along_x(
    mesh=reader.mesh,
    y=0.0,
    z=0.0,
)
```

The vtk file already has the coordinates in inertial frame of the field.
